```python
# The Lost Treasure Game

print("=" * 45)
print("        THE LOST TREASURE")
print("=" * 45)

print("\nYou are an explorer searching for a legendary treasure.")
print("You have entered a mysterious cave.")
print("Your goal is to find the treasure and escape safely!")

score = 0
```

```python
print("\nYou are standing in front of three paths:")
print("1. A dark tunnel")
print("2. A tunnel with a small light")
print("3. A narrow tunnel with strange sounds")

choice = input("\nWhich path do you choose? (1/2/3): ")

if choice == "1":
    print("\nYou enter the dark tunnel...")
    print("You find an old golden key on the ground!")
    score += 20

elif choice == "2":
    print("\nYou follow the light...")
    print("You find a map showing the location of the treasure!")
    score += 30

elif choice == "3":
    print("\nYou follow the strange sounds...")
    print("A group of bats scares you, but you find an ancient coin!")
    score += 10

else:
    print("\nInvalid choice! You lost your chance.")
    score -= 10
```

```python
print("\nYou are standing in front of three paths:")
print("1. A dark tunnel")
print("2. A tunnel with a small light")
print("3. A narrow tunnel with strange sounds")

choice = input("\nWhich path do you choose? (1/2/3): ")

if choice == "1":
    print("\nYou enter the dark tunnel...")
    print("You find an old golden key on the ground!")
    score += 20

elif choice == "2":
    print("\nYou follow the light...")
    print("You find a map showing the location of the treasure!")
    score += 30

elif choice == "3":
    print("\nYou follow the strange sounds...")
    print("A group of bats scares you, but you find an ancient coin!")
    score += 10

else:
    print("\nInvalid choice! You lost your chance.")
    score -= 10
```

```python
print("\n" + "-" * 45)
print("You continue deeper into the cave.")
print("Suddenly, you find a locked wooden door.")

print("There are two ways to open it:")
print("1. Try to break the door")
print("2. Search for a key")

choice = input("\nWhat do you do? (1/2): ")

if choice == "1":
    print("\nYou try to break the door...")

    if score >= 20:
        print("You successfully break the door!")
        score += 20
    else:
        print("The door is too strong. You hurt yourself.")
        score -= 20

elif choice == "2":
    print("\nYou search around the cave...")

    if score >= 20:
        print("You find a hidden key and open the door!")
        score += 30
    else:
        print("You couldn't find the key.")
        score -= 10

else:
    print("\nYou wasted too much time.")
    score -= 10
    
```

```python
print("\n" + "-" * 45)
print("Behind the door, you see a huge treasure chest!")
print("But there is a final challenge.")

print("The chest has three symbols on it:")
print("1. Moon")
print("2. Sun")
print("3. Star")

choice = input("\nWhich symbol do you choose? (1/2/3): ")

if choice == "1":
    print("\nThe chest starts shaking...")
    print("Wrong symbol! The chest remains locked.")
    score -= 20

elif choice == "2":
    print("\nThe chest opens!")
    print("You found the legendary treasure!")
    score += 50

elif choice == "3":
    print("\nThe chest opens halfway...")
    print("You found some gold coins!")
    score += 20

else:
    print("\nYou chose an invalid symbol.")
    score -= 10
```

```python
print("\n" + "=" * 45)
print("             FINAL RESULT")
print("=" * 45)

print(f"Your final score: {score}")

if score >= 70:
    print("AMAZING! You found the legendary treasure!")
    print("You are a true explorer!")

elif score >= 40:
    print("GREAT JOB! You found valuable treasure!")
    print("You escaped the cave successfully!")

elif score >= 0:
    print("You survived the adventure,")
    print("but you did not find the legendary treasure.")

else:
    print("GAME OVER!")
    print("The cave was too dangerous. Better luck next time!")

print("\nThanks for playing The Lost Treasure!")
```
